# Finding Nemo — human-reviewed dataset training

Trains `yolo11n`, `yolo11m`, `yolo26n`, and `yolo26m` in order on the human-reviewed dataset. Each model uses its own local and Camber folder.

- Local: `runs/finding_nemo/<model>/`
- Remote: `stash://bunpmc/projects/finding_nemo/<model>/`
- `last.pt` is saved every epoch; Camber checkpoint sync runs every 5 epochs and whenever a cell is interrupted.
- Re-run a model cell to resume from its local `weights/last.pt`. Completed models are skipped.

Use the repository `.venv` kernel. Run cells top-to-bottom, or use **Run All** for the full sequential suite.

```python
from pathlib import Path
import json
import sys
import time

cwd = Path.cwd().resolve()
if (cwd / 'benchmakring_transmission_line_defect_model').is_dir():
    repo = cwd
elif cwd.name == 'benchmakring_transmission_line_defect_model':
    repo = cwd.parent
else:
    candidates = [p for p in [cwd, *cwd.parents] if (p / 'benchmakring_transmission_line_defect_model').is_dir()]
    if not candidates:
        raise RuntimeError('Run this notebook from the repository checkout')
    repo = candidates[0]

script_dir = repo / 'benchmakring_transmission_line_defect_model/scripts'
if str(script_dir) not in sys.path:
    sys.path.insert(0, str(script_dir))

import train_finding_nemo as nemo

state = nemo.prepare_state()
print('Dataset:', nemo.DATA)
print('GPU:', state['device'])
print('Camber:', state['remote_root'])
print(json.dumps(state.get('runs', {}), indent=2))
```

## Inspect progress without training

This cell is safe to run at any time.

```python
current = nemo.read_state()
for model_name in nemo.ORDER:
    run = current.get('runs', {}).get(model_name, {})
    last = nemo.RUNS / model_name / 'weights/last.pt'
    results = nemo.RUNS / model_name / 'results.csv'
    completed_epochs = max(0, len(results.read_text(encoding='utf-8').splitlines()) - 1) if results.is_file() else 0
    print(f'{model_name:8} status={run.get("status", "pending"):12} epochs={completed_epochs:2}/50 last.pt={last.is_file()}')
```

```python
benchmark_dir = nemo.BENCHMARK
if str(benchmark_dir) not in sys.path:
    sys.path.insert(0, str(benchmark_dir))

import importlib
import src.evaluator as evaluator_module
evaluator_module = importlib.reload(evaluator_module)
BenchmarkEvaluator = evaluator_module.BenchmarkEvaluator

model_names = ('yolo11n', 'yolo11m', 'yolo26n', 'yolo26m', 'rtdetr50', 'rtdetr101')
checkpoint_paths = {
    model_name: nemo.RUNS / model_name / 'weights' / 'best.pt'
    for model_name in model_names
}
missing_checkpoints = [str(path) for path in checkpoint_paths.values() if not path.is_file()]
if missing_checkpoints:
    raise FileNotFoundError('Missing best.pt checkpoints:\n' + '\n'.join(missing_checkpoints))
if not nemo.DATA.is_file():
    raise FileNotFoundError(f'Finding Nemo dataset YAML not found: {nemo.DATA}')

evaluation_dir = nemo.RUNS / 'test_evaluation'
evaluation_dir.mkdir(parents=True, exist_ok=True)
evaluator = BenchmarkEvaluator()
evaluator.results_dir = evaluation_dir
evaluator.eval_file = evaluation_dir / 'benchmark_test_results.json'
evaluator.eval_data = {}
evaluation_model_names = {'rtdetr101': 'rtdetr100'}
test_runs = {
    f'nemo_{evaluation_model_names.get(model_name, model_name)}': {'best_weights': str(checkpoint_path)}
    for model_name, checkpoint_path in checkpoint_paths.items()
}

print(f'Evaluating Finding Nemo best.pt checkpoints on the test split: {nemo.DATA}')
df_test_all = evaluator.evaluate_all(
    test_runs,
    nemo.DATA,
    split_by_dataset={'nemo': 'test'},
    checkpoint_field='best_weights',
)
df_test_all.loc[df_test_all['Model'] == 'rtdetr100', 'Model'] = 'rtdetr101'
metric_columns = [
    'Model', 'Evaluation Split', 'Precision', 'Recall', 'F1-Score',
    'mAP@0.5', 'mAP@0.5:0.95', 'MEAN_5', 'Latency (ms)', 'FPS',
    'GFLOPS', 'Params (M)', 'VRAM (MB)',
]
df_test = df_test_all.loc[:, metric_columns]
df_test.to_csv(evaluation_dir / 'test_metrics.csv', index=False)
display(df_test)
```
